```python
import math
```

```python
x = [0, 1, 2, 3, 4]
p = [0.15, 0.35, 0.25, 0.15, 0.10]
```

```python
esperanza = 0
for i in range(len(x)):
    esperanza += x[i] * p[i]
```

```python
esperanza_cuadrado = 0
for i in range(len(x)):
    esperanza_cuadrado += (x[i] ** 2) * p[i]
```

```python
varianza = esperanza_cuadrado - (esperanza ** 2)
desviacion = math.sqrt(varianza)
```
