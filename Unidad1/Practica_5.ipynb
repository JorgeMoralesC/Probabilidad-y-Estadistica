```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
covid = pd.read_csv("covid.csv", encoding = "ISO-8859-1")
```

```python
covid.head()
```

```python
covid.info()
```

```python
covid.EDAD.max()
```

```python
covid.EDAD.min()
```

```python
covid.EDAD.describe()
```

```python
conteo_edades = covid['EDAD'].value_counts().sort_index()
plt.bar(conteo_edades.index, conteo_edades.values, color='skyblue',edgecolor='black', alpha=0.6)
plt.xlabel('Edad')
plt.title("Pruebas covid por edades")
plt.ylabel('Frecuencia')
plt.show()
```

```python
media = covid.EDAD.mean()
rangos = [0,10,20,30,40,50,60,80,90,100,120]
plt.hist(covid.EDAD, bins=rangos, edgecolor='black')
plt.title("Histograma de pruebas por covid")
plt.axvline(media,color='y',linestyle='dashed',linewidth=2,label=f'Media Edad(media:.2f)')
plt.xlabel('Edad')
plt.ylabel('Frecuencia')
plt.legend()
```

```python
conteo_sexo = covid['SEXO'].value_counts()
plt.bar(conteo_sexo.index, conteo_sexo.values, color='skyblue',edgecolor='black', alpha=0.7)
plt.xlabel('Sexo')
plt.title("Pruebas covid por sexo")
plt.ylabel('Frecuencia')
```

```python
conteo_tipo_paciente = covid['TIPO_PACIENTE'].value_counts()
plt.bar(conteo_tipo_paciente.index, conteo_tipo_paciente.values, color='skyblue',edgecolor='black', alpha=0.7)
plt.xlabel('Paciente')
plt.title("Pruebas covid por pacientes")
plt.ylabel('Frecuencia')
```

```python
sns.set_style("whitegrid")
sns.boxplot(x="SEXO", y="EDAD", data=covid, palette="pastel")

plt.title("distribucion de edad por sexo", fontsize=14, fontweight='bold')
plt.xlabel("Sexo", fontsize=12)
plt.ylabel("Edad", fontsize=12)
plt.show()
```
