```python
tipos = carro["body_type"].dropna().unique()
```

```python
datos = []

for tipo in tipos:
    precios = carro[carro["body_type"] == tipo]["price_eur"].dropna()
    datos.append(precios)
```

```python
plt.figure(figsize=(12, 6))

plt.boxplot(datos, tick_labels=tipos)

plt.title("Relación entre el precio y el tipo de vehículo")
plt.xlabel("Tipo de vehículo")
plt.ylabel("Precio del automóvil (EUR)")

plt.xticks(rotation=25)

plt.show()
```

```python
tracciones = carro["drive_type"].dropna().unique()
```

```python
datos = []

for traccion in tracciones:
    precios = carro[carro["drive_type"] == traccion]["price_eur"].dropna()
    datos.append(precios)
```

```python
plt.figure(figsize=(10, 6))

plt.boxplot(datos, tick_labels=tracciones)

plt.title("Relación entre el precio y el tipo de tracción")
plt.xlabel("Tipo de tracción")
plt.ylabel("Precio del automóvil (EUR)")

plt.show()
```
