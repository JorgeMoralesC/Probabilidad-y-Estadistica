```python
import math as mt
from itertools import permutations
```

```python
asistentes = [f"Asistente_{i}" for i in range(1, 11)]
```

```python
n = len(asistentes)
r = 4
```

```python
Pn = mt.factorial(n) / mt.factorial(n - r)
```

```python
print(f"Número total de maneras de elegir asistentes: {Pn:.0f}")
```

```python
print("\nGenerando las permutaciones de selección:")
permutaciones_asistentes = list(permutations(asistentes, r))
```

```python
for i, perm in enumerate(permutaciones_asistentes[:10], 1):
    print(f"Opción {i}: {perm}")
```

```python
print(f"\n... y {len(permutaciones_asistentes) - 10} opciones más.")
```
